## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python

marks = []

for i in range(5):
    num = int(input("Enter marks: "))

    if num == -1:
        break

    if num < 0 or num > 100:
        print("Invalid marks")
        continue

    marks.append(num)

print("Valid Marks:", marks)

if len(marks) > 0:
    average = sum(marks) / len(marks)
    print("Average =", average)
else:
    print("No valid marks entered")
```

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python

rooms = [101, 102, 103, 104, 105]
booked = []

while len(booked) < 5:
    print("Available rooms:", rooms)
    room = int(input("Choose room: "))

    if room not in rooms:
        print("Room not available.")
        continue

    rooms.remove(room)
    booked.append(room)
    print("Room", room, "booked successfully!")

print("All rooms are booked.")
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python

balance = 0

while True:
    print("\n1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("Account created successfully!")

    elif choice == "2":5
        amount = int(input("Enter deposit amount: "))
        balance = balance + amount
        print("Amount deposited successfully!")

    elif choice == "3":
        amount = int(input("Enter withdraw amount: "))

        if amount <= balance:
            balance = balance - amount
            print("Amount withdrawn successfully!")
        else:
            print("Insufficient balance!")

    elif choice == "4":
        print("Your balance is:", balance)

    elif choice == "5":
        print("Thank you for using our bank!")
        break

    else:
        print("Invalid choice!")
```

## School Marks Report

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python

for student in range(1, 6):
    total = 0
    print("Student", student)

    for subject in range(1, 5):
        marks = int(input("Enter marks for Subject " + str(subject) + ": "))
        total = total + marks

        if marks < 33:
            print("Failed in Subject", subject)

    percentage = total / 4

    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 33:
        grade = "D"
    else:
        grade = "F"

    print("Total Marks =", total)
    print("Percentage =", percentage, "%")
    print("Grade =", grade)
    print()
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python

for package in range(1, 4):
    total = 0
    count = 0

    print("Package", package)

    for feature in range(1, 4):
        rating = int(input("Enter rating out of 5: "))

        if rating == 0:
            continue

        total = total + rating
        count = count + 1

    if count > 0:
        average = total / count
        print("Average Rating =", round(average, 2))
    else:
        print("No ratings available")

    print()
```
